### Contar valores únicos em todas as colunas
Esta função retorna a quantidade de valores únicos e os próprios valores únicos para cada coluna do DataFrame.

```python
def contar_valores_unicos_todas_colunas(df):
    resultado = {}
    for col in df.columns:
        n_unicos = df[col].nunique()
        valores_unicos = df[col].unique()
        resultado[col] = {'quantidade': n_unicos, 'valores': valores_unicos}
    return resultado

# Exemplo de uso:
contagem_unicos = contar_valores_unicos_todas_colunas(df_sp)
for coluna, info in contagem_unicos.items():
    print(f'{coluna}: {info["quantidade"]} valores únicos')
```

### Função para contar valores 'DESCONHECIDO' ou 'NAO INFORMADO' em todas as colunas
Esta função percorre todas as colunas do DataFrame e retorna a contagem desses valores para cada coluna.

```python
def contar_desconhecido_ou_nao_informado(df):
    resultado = {}
    for col in df.columns:
        count = df[col].isin(['DESCONHECIDO', 'NAO INFORMADO']).sum()
        if count > 0:
            resultado[col] = count
    return resultado

# Exemplo de uso:
contagem = contar_desconhecido_ou_nao_informado(df_sp)
print('Contagem de "DESCONHECIDO" ou "NAO INFORMADO" por coluna:')
for coluna, qtd in contagem.items():
    print(f'{coluna}: {qtd}')

contagem = contar_desconhecido_ou_nao_informado(df_sp)
```

## Adicionando o tipo de veículo ao dataset de acidentes

Vamos adicionar uma coluna com o(s) tipo(s) de veículo(s) envolvidos em cada acidente, utilizando a tabela de tipos de veículos e relacionando pelo campo `num_acidente`.

```python
# Carregar o dataset de tipos de veículos
veic_path = 'dataset/renaest_dabertos_20250412/TipoVeiculo_DadosAbertos_20250412.csv'
df_veic = pd.read_csv(veic_path, sep=';', encoding='utf-8', low_memory=False)

# Se houver mais de um tipo de veículo por acidente, agregue em uma string separada por vírgula
veic_por_acidente = df_veic.groupby('num_acidente')['tipo_veiculo'].apply(lambda x: ', '.join(sorted(x.unique()))).reset_index()

# Adiciona a coluna ao dataset principal (df_sp_limpo)
df_sp_limpo = df_sp_limpo.merge(veic_por_acidente, on='num_acidente', how='left')

# Exibe amostra para conferência
display(df_sp_limpo[['num_acidente', 'tipo_veiculo']].head())
```

```python
#analisar os dados após a junção
df_sp_limpo.info()
df_sp_limpo.describe()

#identificar quantidade de valores faltantes
valores_faltantes = df_sp_limpo.isnull().sum()
print(valores_faltantes[valores_faltantes > 0])
#remover colunas que não são mais necessárias

#Verificar presença de valores DESCONHECIDO ou NAO INFORMADO na coluna tipo_veiculo
#Vamos contar quantos acidentes tem os valores Desconhecido ou Nao Informado na coluna tipo_veiculo
contagem = df_sp_limpo['tipo_veiculo'].isin(['DESCONHECIDO', 'NAO INFORMADO']).sum()
print(f"Quantidade de acidentes com tipo de veículo 'DESCONHECIDO' ou 'NAO INFORMADO': {contagem}")

#Como a quantidade de registros desconhecidos ou não informados é baixa, podemos remover esses registros
df_sp_limpo = df_sp_limpo[~df_sp_limpo['tipo_veiculo'].isin(['DESCONHECIDO', 'NAO INFORMADO'])]

# Verificar a quantidade de acidentes por tipo de veículo
motocicleta = df_sp_limpo['tipo_veiculo'].str.contains('MOTOCICLETA', na=False).sum()
automovel = df_sp_limpo['tipo_veiculo'].str.contains('AUTOMOVEL', na=False).sum()
onibus = df_sp_limpo['tipo_veiculo'].str.contains('ONIBUS', na=False).sum()
bicicleta = df_sp_limpo['tipo_veiculo'].str.contains('BICICLETA', na=False).sum()
pedestre = df_sp_limpo['tipo_veiculo'].str.contains('PEDESTRE', na=False).sum()

print(f"Quantidade de acidentes com motocicleta: {motocicleta}")
print(f"Quantidade de acidentes com automóvel: {automovel}")
print(f"Quantidade de acidentes com ônibus: {onibus}")
print(f"Quantidade de acidentes com bicicleta: {bicicleta}")
print(f"Quantidade de acidentes envolvendo pedestre: {pedestre}")
```

## Distribuição das Vítimas por Tipo de Usuário da Via

Quanto ao tipo de usuário da via, [Percentual]% das vítimas eram [Tipo, ex: motociclistas], seguidos por [Tipo, ex: pedestres] ([Percentual]%) e [Tipo, ex: ocupantes de automóveis] ([Percentual]%) (Figura D). A alta proporção de [mencionar o grupo mais vulnerável] destaca a fragilidade deste grupo no contexto urbano paulistano.

```python
# Ajuste o nome da coluna conforme o dataset (ex: 'tp_envolvido', 'tipo_usuario', etc.)
col_usuario = 'tipo_veiculo'

# Considera apenas vítimas fatais, se disponível
if 'gravidade_lesao' in df_sp_limpo.columns:
    vitimas_fatais = df_sp_limpo[df_sp_limpo['gravidade_lesao'].str.upper().str.contains('OBITO') | df_sp_limpo['gravidade_lesao'].str.upper().str.contains('OBITO')]
else:
    vitimas_fatais = df_sp_limpo.copy()

usuarios = vitimas_fatais[col_usuario].value_counts()
total = usuarios.sum()
percentuais = (usuarios / total * 100).round(1)

motocicleta = df_sp_limpo['tipo_veiculo'].str.contains('MOTOCICLETA', na=False).sum()
automovel = df_sp_limpo['tipo_veiculo'].str.contains('AUTOMOVEL', na=False).sum()
onibus = df_sp_limpo['tipo_veiculo'].str.contains('ONIBUS', na=False).sum()
bicicleta = df_sp_limpo['tipo_veiculo'].str.contains('BICICLETA', na=False).sum()
pedestre = df_sp_limpo['tipo_veiculo'].str.contains('PEDESTRE', na=False).sum()


# Seleciona os três principais tipos
top3 = percentuais.head(3)

# Identifica o grupo mais vulnerável
grupo_mais_vulneravel = top3.index[0].lower()

print(
    f"Quanto ao tipo de usuário da via, {top3.iloc[0]}% das vítimas eram {top3.index[0].lower()}, "
    f"seguidos por {top3.index[1].lower()} ({top3.iloc[1]}%) e {top3.index[2].lower()} ({top3.iloc[2]}%) (Figura D). "
    f"A alta proporção de {grupo_mais_vulneravel} destaca a fragilidade deste grupo no contexto urbano paulistano."
)

# Gráfico
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))
usuarios.plot(kind='bar', color='mediumseagreen')
plt.title('Óbitos por Tipo de Usuário da Via')
plt.xlabel('Tipo de Usuário')
plt.ylabel('Quantidade de Óbitos')
plt.tight_layout()
plt.show()
```
